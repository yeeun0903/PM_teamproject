```python
import pandas as pd
import plotly.express as px

df = pd.read_excel("coupangplay_dataproject_survey_179.xlsx")

cols = [
    "7. 쿠팡플레이는 짧은 시간에 보기 좋은 콘텐츠가 부족하다고 생각한다.",
    "15. 시간이 부족해 OTT 콘텐츠를 자주 시청하지 못한다.",
    "14. 한 편의 길이가 길어 끝까지 보기 부담스럽다.",
    "12. 나는 긴 영상보다 짧은 영상을 더 부담 없이 시청한다.",
    "16. 쿠팡플레이 숏폼 드라마·예능을 시청할 의향이 있다."
]

label_map = {
    cols[0]: "짧은 콘텐츠 부족",
    cols[1]: "시간 부족",
    cols[2]: "길이 부담",
    cols[3]: "짧은 영상 선호",
    cols[4]: "숏폼 시청 의향"
}

df_num = df[cols].apply(pd.to_numeric, errors="coerce")
df_num = df_num.rename(columns=label_map)

corr_matrix = df_num.corr()

fig = px.imshow(
    corr_matrix,
    text_auto=".2f",
    color_continuous_scale="Blues",
    zmin=-1,
    zmax=1,
    title="콘텐츠 길이 요인과 숏폼 시청 의향 간 상관관계"
)

fig.update_layout(
    width=700,
    height=600
)

fig.show()
```
